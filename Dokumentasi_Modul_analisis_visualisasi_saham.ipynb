## **Langkah-Langkah**

### **1. Mengimpor Library**
Kode berikut digunakan untuk mengimpor library yang dibutuhkan:


```python
import numpy as np  
import pandas as pd  
from pylab import plt, mpl  

# Mengatur gaya visualisasi
plt.style.use('ggplot')  
mpl.rcParams['font.family'] = 'serif'  
%matplotlib inline
```

- **`numpy`**: Untuk operasi numerik.
- **`pandas`**: Untuk manipulasi data dalam bentuk tabel.
- **`pylab`**: Untuk plotting dan operasi numerik.
- **`plt.style.use('ggplot')`**: Mengatur gaya visualisasi grafik.
- **`mpl.rcParams['font.family'] = 'serif'`**: Mengatur font grafik menjadi serif.
- **`%matplotlib inline`**: Menampilkan grafik langsung di notebook.

---

### **2. Membaca Data**
Kode berikut digunakan untuk membaca data dari file CSV atau membuat dataset dummy jika file tidak ditemukan:


```python
try:
    # Membaca data dari file CSV
    data = pd.read_csv('../../source/tr_eikon_eod_data.csv',
                       index_col=0, parse_dates=True)
except FileNotFoundError:
    # Membuat dataset dummy jika file tidak ditemukan
    dates = pd.date_range(start='2010-01-01', end='2024-06-29', freq='B')
    spx_values = np.cumprod(1 + np.random.normal(0, 0.001, len(dates))) * 1000
    data = pd.DataFrame(spx_values, index=dates, columns=['.SPX'])
```

- **`pd.read_csv`**: Membaca file CSV.
- **`pd.date_range`**: Membuat rentang tanggal.
- **`np.random.normal`**: Membuat data acak dengan distribusi normal.
- **`np.cumprod`**: Menghitung hasil kumulatif (simulasi harga saham).

---

### **3. Membersihkan Data**
Kode berikut digunakan untuk membersihkan data:


```python
data = pd.DataFrame(data['.SPX']) 
data.dropna(inplace=True)  
data.info()
```

- **`data['.SPX']`**: Memastikan hanya kolom harga saham yang digunakan.
- **`dropna(inplace=True)`**: Menghapus baris dengan nilai kosong.
- **`data.info()`**: Menampilkan informasi tentang DataFrame.

---

### **4. Menghitung Return dan Volatilitas**
#### **4.1 Menghitung Log Return**


```python
data['rets'] = np.log(data / data.shift(1))
```

- **`data.shift(1)`**: Menggeser data satu hari ke belakang.
- **`np.log(data / data.shift(1))`**: Menghitung log return harian.

#### **4.2 Menghitung Volatilitas**


```python
data['vola'] = data['rets'].rolling(252).std() * np.sqrt(252)
```

- **`rolling(252).std()`**: Menghitung standar deviasi rolling selama 252 hari (1 tahun kerja).
- **`np.sqrt(252)`**: Mengonversi volatilitas harian menjadi tahunan.

---

### **5. Visualisasi Data**
Kode berikut digunakan untuk memvisualisasikan data harga saham dan volatilitas:


```python
data[['.SPX', 'vola']].plot(subplots=True, figsize=(10, 6));
```

- **`data[['.SPX', 'vola']]`**: Memilih kolom harga saham dan volatilitas.
- **`plot(subplots=True)`**: Membuat dua grafik terpisah.
- **`figsize=(10, 6)`**: Mengatur ukuran grafik.
